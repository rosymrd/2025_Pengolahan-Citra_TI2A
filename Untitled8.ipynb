```python
import matplotlib.pyplot as plt
from skimage import data, filters

# 1. Memuat citra (contoh: coins)
image_coins = data.coins()  # Citra sudah dalam format grayscale

# 2. Thresholding Global Manual
thresh_manual = 100
binary_manual = image_coins > thresh_manual

# 3. Thresholding dengan Metode Otsu
thresh_otsu = filters.threshold_otsu(image_coins)
binary_otsu = image_coins > thresh_otsu

# 4. Visualisasi Hasil
fig, axes = plt.subplots(ncols=3, figsize=(12, 4))
ax = axes.ravel()

ax[0].imshow(image_coins, cmap='gray')
ax[0].set_title('Original Image')
ax[0].axis('off')

ax[1].imshow(binary_manual, cmap='gray')
ax[1].set_title(f'Manual Threshold (T={thresh_manual})')
ax[1].axis('off')

ax[2].imshow(binary_otsu, cmap='gray')
ax[2].set_title(f"Otsu's Threshold (T={thresh_otsu:.2f})")
ax[2].axis('off')

plt.tight_layout()
plt.show()

# Menampilkan nilai threshold Otsu
print(f"Nilai threshold Otsu yang ditemukan: {thresh_otsu}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, segmentation

# 1. Memuat citra (contoh: camera)
image_camera = data.camera()

# 2. Menentukan titik seed (benih)
seed_point = (50, 150)  # Koordinat y=50, x=150

# 3. Melakukan flood fill (region growing) dengan toleransi perbedaan intensitas
flood_mask = segmentation.flood(image_camera, seed_point, tolerance=10)

# 4. Membuat citra hasil segmentasi dengan menandai area hasil flood fill
segmented_image = np.copy(image_camera)
segmented_image[flood_mask] = 255  # Tandai area dengan warna putih

# 5. Visualisasi hasil segmentasi
fig, axes = plt.subplots(ncols=3, figsize=(12, 4))
ax = axes.ravel()

# Gambar asli dengan titik seed
ax[0].imshow(image_camera, cmap='gray')
ax[0].plot(seed_point[1], seed_point[0], 'ro')  # Tandai titik seed
ax[0].set_title('Original Image with Seed')
ax[0].axis('off')

# Mask hasil flood fill
ax[1].imshow(flood_mask, cmap='gray')
ax[1].set_title('Flood Fill Mask (Region)')
ax[1].axis('off')

# Citra hasil segmentasi
ax[2].imshow(segmented_image, cmap='gray')
ax[2].set_title('Segmented Image (Region Marked)')
ax[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage.color import rgb2lab, lab2rgb
from sklearn.cluster import KMeans
import warnings

# 1. Memuat citra warna (contoh: astronaut)
image = data.astronaut()

# 2. Konversi ke float dan ubah ke ruang warna Lab
image_float = image.astype(float) / 255.0
image_lab = rgb2lab(image_float)

# 3. Ubah citra ke array piksel [jumlah_piksel, jumlah_fitur]
rows, cols, channels = image_lab.shape
pixels = image_lab.reshape(rows * cols, channels)

# 4. Terapkan K-Means Clustering
n_clusters = 4
kmeans = KMeans(n_clusters=n_clusters, random_state=0, n_init=10)

# Hindari warning versi baru sklearn
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    labels = kmeans.fit_predict(pixels)

# 5. Bentuk ulang label ke dimensi citra
label_image = labels.reshape(rows, cols)

# 6. Buat citra segmentasi dengan warna pusat klaster
segmented_lab = np.zeros_like(image_lab)
for i in range(n_clusters):
    mask = (labels == i).reshape(rows, cols)
    segmented_lab[mask] = kmeans.cluster_centers_[i]

# 7. Konversi kembali ke RGB
segmented_rgb = lab2rgb(segmented_lab)

# 8. Visualisasi hasil
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
ax = axes.ravel()

ax[0].imshow(image)
ax[0].set_title('Original Image')
ax[0].axis('off')

ax[1].imshow(label_image, cmap='viridis')
ax[1].set_title(f'Cluster Labels (K={n_clusters})')
ax[1].axis('off')

ax[2].imshow(segmented_rgb)
ax[2].set_title(f'Segmented Image (K={n_clusters})')
ax[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, filters, segmentation, morphology, measure
from scipy import ndimage as ndi
from skimage.color import label2rgb

# 1. Memuat citra grayscale (contoh: coins)
image = data.coins()

# 2. Hitung gradien sebagai 'topografi'
elevation = filters.sobel(image)

# 3. Marker sederhana menggunakan thresholding
markers = np.zeros_like(image)
markers[image < 30] = 1   # latar belakang
markers[image > 150] = 2  # objek (koin)

# 4. Terapkan algoritma watershed
segmentation_result = segmentation.watershed(elevation, markers)

# 5. Tambahkan batas kuning pada hasil segmentasi
segmented_display = segmentation.mark_boundaries(image, segmentation_result, color=(1, 1, 0))  # batas kuning

# 6. Visualisasi
fig, axes = plt.subplots(1, 4, figsize=(18, 5), sharex=True, sharey=True)
titles = ['Original Image', 'Elevation Map (Sobel Gradient)', 'Markers', 'Watershed Segmentation']
images = [image, elevation, markers, segmented_display]
cmaps = [plt.cm.gray, plt.cm.nipy_spectral, plt.cm.gray, None]  # markers dibuat grayscale agar mirip referensi

for ax, img, title, cmap in zip(axes, images, titles, cmaps):
    ax.imshow(img, cmap=cmap)
    ax.set_title(title)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from skimage import data, filters, segmentation, img_as_float
from sklearn.cluster import KMeans
import numpy as np
import warnings

# 1. Load sample image (grayscale)
image = data.camera()
image_float = img_as_float(image)

# 2a. Otsu Thresholding
threshold_value = filters.threshold_otsu(image)
segmentation_otsu = image > threshold_value

# 2b. K-Means Clustering (K=3)
rows, cols = image.shape
pixels = image_float.reshape(rows * cols, 1)
k = 3
kmeans = KMeans(n_clusters=k, random_state=0, n_init=10)

with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    kmeans_labels = kmeans.fit_predict(pixels)

segmentation_kmeans = kmeans_labels.reshape(rows, cols)

# 2c. Watershed Segmentation (use Otsu for markers)
elevation_map = filters.sobel(image)
markers = np.zeros_like(image)
markers[image < threshold_value] = 1
markers[image > threshold_value] = 2
segmentation_ws = segmentation.watershed(elevation_map, markers)

# Overlay segmentation boundaries on original image
segmentation_ws_borders = segmentation.mark_boundaries(image_float, segmentation_ws, color=(1, 0, 0))  # Red borders

# 3. Plot comparison
fig, axes = plt.subplots(1, 4, figsize=(18, 5), sharex=True, sharey=True)
titles = ['Original Image', 'Otsu Thresholding', f'K-Means (K={k})', 'Watershed']
images = [image, segmentation_otsu, segmentation_kmeans, segmentation_ws_borders]
cmaps = [plt.cm.gray, plt.cm.gray, 'viridis', None]

for ax, img, title, cmap in zip(axes, images, titles, cmaps):
    ax.imshow(img, cmap=cmap)
    ax.set_title(title)
    ax.axis('off')

plt.suptitle('Comparison of Segmentation Methods', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
# Instalasi (jika belum)
!pip install scikit-image ipywidgets --quiet

import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color, filters, segmentation
from skimage.filters import threshold_otsu
from skimage.color import label2rgb
from sklearn.cluster import KMeans
from scipy import ndimage as ndi
from skimage.feature import peak_local_max
from PIL import Image
import io as io_bytes
import ipywidgets as widgets
from IPython.display import display

# Fungsi segmentasi dan visualisasi
def segment_and_display(image_array):
    image_gray = color.rgb2gray(image_array)

    # 1. Otsu Thresholding
    thresh_otsu = threshold_otsu(image_gray)
    segmented_otsu = image_gray > thresh_otsu

    # 2. K-Means Clustering
    pixel_values = image_array.reshape((-1, 3))
    kmeans = KMeans(n_clusters=3, n_init='auto')
    kmeans.fit(pixel_values)
    segmented_kmeans = kmeans.labels_.reshape(image_array.shape[:2])

    # 3. Watershed
    gradient = filters.sobel(image_gray)
    coords = peak_local_max(-gradient, footprint=np.ones((3, 3)))  # FIXED
    local_maxi = np.zeros_like(gradient, dtype=bool)
    local_maxi[tuple(coords.T)] = True
    markers = ndi.label(local_maxi)[0]
    labels_ws = segmentation.watershed(gradient, markers)

    # Visualisasi
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    ax = axes.ravel()

    ax[0].imshow(image_array)
    ax[0].set_title("Original Image")
    ax[0].axis('off')

    ax[1].imshow(segmented_otsu, cmap='gray')
    ax[1].set_title("Otsu Thresholding")
    ax[1].axis('off')

    ax[2].imshow(segmented_kmeans, cmap='viridis')
    ax[2].set_title("K-Means Clustering")
    ax[2].axis('off')

    ax[3].imshow(label2rgb(labels_ws, image=image_array, bg_label=0))
    ax[3].set_title("Watershed Segmentation")
    ax[3].axis('off')

    plt.tight_layout()
    plt.show()

# Widget upload file
uploader = widgets.FileUpload(accept='image/*', multiple=False)

def on_upload_change(change):
    if uploader.value:
        content = list(uploader.value.values())[0]['content']
        image = Image.open(io_bytes.BytesIO(content)).convert('RGB')
        image_array = np.array(image)
        segment_and_display(image_array)

uploader.observe(on_upload_change, names='value')
display(widgets.HTML(value="<h3>Unggah gambar untuk segmentasi (misalnya lemon):</h3>"))
display(uploader)
```

```python
import matplotlib.pyplot as plt
from skimage import data, filters
from skimage.filters import threshold_otsu, threshold_local, threshold_yen
from skimage.color import rgb2gray

# 1. Memuat citra data.page() (sudah grayscale)
image_page = data.page()

# 2. Thresholding dengan Otsu
thresh_otsu = threshold_otsu(image_page)
binary_otsu = image_page > thresh_otsu

# 3. Thresholding dengan Local Threshold
block_size = 35  # harus ganjil
local_thresh = threshold_local(image_page, block_size, offset=10)
binary_local = image_page > local_thresh

# 4. Thresholding dengan Yen
thresh_yen = threshold_yen(image_page)
binary_yen = image_page > thresh_yen

# 5. Visualisasi hasil
fig, axes = plt.subplots(nrows=1, ncols=4, figsize=(16, 5))
ax = axes.ravel()

ax[0].imshow(image_page, cmap='gray')
ax[0].set_title('Original Image')
ax[0].axis('off')

ax[1].imshow(binary_otsu, cmap='gray')
ax[1].set_title(f"Otsu's Threshold\n(T={thresh_otsu:.2f})")
ax[1].axis('off')

ax[2].imshow(binary_local, cmap='gray')
ax[2].set_title(f"Local Threshold\n(block={block_size})")
ax[2].axis('off')

ax[3].imshow(binary_yen, cmap='gray')
ax[3].set_title(f"Yen's Threshold\n(T={thresh_yen:.2f})")
ax[3].axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, segmentation, color
from skimage.color import rgb2gray

# Gunakan gambar contoh dari skimage
image_rgb = data.camera()  # Jika ingin gambar lain, bisa diganti
image_gray = image_rgb  # Sudah grayscale

# Definisikan titik seed untuk eksperimen
seed_point = (10, 100)  # Misalnya di langit (ubah sesuai kebutuhan)

# Coba dua nilai tolerance
for tol in [5, 50]:
    # Aplikasikan flood fill
    flood_mask = segmentation.flood(image_gray, seed_point, tolerance=tol)

    # Tandai hasil segmentasi
    segmented_image = np.copy(image_gray)
    segmented_image[flood_mask] = 255  # Tandai area yang tumbuh

    # Tampilkan hasilnya
    fig, axes = plt.subplots(ncols=3, figsize=(12, 4))
    ax = axes.ravel()

    ax[0].imshow(image_gray, cmap='gray')
    ax[0].plot(seed_point[1], seed_point[0], 'ro')
    ax[0].set_title(f'Original with Seed\nTolerance={tol}')
    ax[0].axis('off')

    ax[1].imshow(flood_mask, cmap='gray')
    ax[1].set_title('Flood Fill Mask')
    ax[1].axis('off')

    ax[2].imshow(segmented_image, cmap='gray')
    ax[2].set_title('Segmented Image')
    ax[2].axis('off')

    plt.tight_layout()
    plt.show()
```
